### №23. Берілген CDF арқылы ықтималдықтарды есептеу
$X$ — кездейсоқ таңдалған автомобильдегі елеулі ақаулар саны. Оның CDF-і:

$$
F(x)=\begin{cases}
0, & x<0\\
0.06, & 0\le x<1\\
0.19, & 1\le x<2\\
0.39, & 2\le x<3\\
0.67, & 3\le x<4\\
0.92, & 4\le x<5\\
0.97, & 5\le x<6\\
1, & 6\le x.
\end{cases}
$$

CDF-ті пайдаланып есептеңіз:  
a. $P(X=2)$  
b. $P(X>3)$  
c. $P(2\le X\le5)$  
d. $P(2<X<5)$.

### №24. CDF → PMF
Кездейсоқ таңдалған полис иесі үшін $X$ — кезекті екі сақтандыру төлемінің арасындағы айлар саны. CDF:

$$
F(x)=\begin{cases}
0, & x<1\\
0.30, & 1\le x<3\\
0.40, & 3\le x<4\\
0.45, & 4\le x<6\\
0.60, & 6\le x<12\\
1, & 12\le x.
\end{cases}
$$

a. $X$-тің PMF-ын табыңыз.  
b. Тек CDF-ті пайдаланып $P(3\le X\le6)$ және $P(4\le X)$ есептеңіз.

## Bernoulli Distribution

Тиын лақтыруды 0 мен 1 аралығындағы кездейсоқ сан арқылы модельдеуге болады: 0.5-тен жоғары мәнді «елтаңба», ал төмен мәнді «сан» деп аламыз.


```python
np.random.random()
```

Осы әдіспен 100 рет тиын лақтыруды модельдейік.


```python
#Note that last week I was using the 'append' function to add to our list of simulated trials, but the below is actually better practice.
random_numbers = np.empty(100)
for i in range(100):
    x = np.random.random()
    if x >= 0.5:
        random_numbers[i] = 1
    else:
        random_numbers[i] = 0
plt.hist(random_numbers)
```

Модельдеуді `scipy.stats` модулі арқылы да орындауға болады.


```python
dist = stats.bernoulli(0.5)
random_numbers = dist.rvs(size=100)
plt.hist(random_numbers)
```

Үлестірім объектісі арқылы орташа мәнді, дисперсияны, PMF және CDF мәндерін есептеуге болады.


```python
# Орташа мән
dist.mean()
```

```python
# Дисперсия
dist.var()
```

```python
# 0 мәні үшін PMF
dist.pmf(0)
```

```python
# 1 мәні үшін PMF
dist.pmf(1)
```

```python
# 0 мәні үшін CDF
dist.cdf(0)
```

```python
# 1 мәні үшін CDF
dist.cdf(1)
```

## Биномдық үлестірім


Биномдық үлестірімді 100 рет тәуелсіз тиын лақтыру мысалында қарастырайық.


```python
dist = stats.binom(n=100, p=0.5)
```

Төмендегі код 100 тиынды лақтыру тәжірибесін 100 рет модельдейді және әр тәжірибедегі сәттіліктер санын береді.


```python
dist.rvs(size=100)
```

```python
plt.hist(dist.rvs(size=100))
```

```python
# Орташа мән
dist.mean()
```

```python
# Дисперсия
dist.var()
```

```python
# 50 мәні үшін PMF
dist.pmf(50)
```

```python
# 50 мәні үшін CDF
dist.cdf(50)
```

```python
# CDF-ті қолмен есептеу
x = 0
for i in range(51):
    x += dist.pmf(i)
x
```

Егер 100 емес, тек 3 сынақ жүргізсек, мүмкін болатын әрбір сәттілік саны үшін PMF мәндерін көрсетуге болады.


```python
values = np.array([])
pmf_values = np.array([])
for i in range(4):
    values = np.append(values, i)
    pmf_values = np.append(pmf_values, stats.binom.pmf(n=3, p=0.5, k=i))
plt.hist(values, weights=pmf_values)
```

## Matplotlib көмегімен графиктерді дұрыс рәсімдеу


Графиктің өлшемін `figsize` параметрі арқылы үлкейтуге болады.


```python
# Гистограмма: figsize график өлшемін үлкейтеді
fig = plt.figure(figsize=(10,10))
plt.hist(values, weights=pmf_values)
```

x осіндегі белгілерді (`ticks`) баптай аламыз.


```python
fig = plt.figure(figsize=(10,10))
plt.xticks(range(4))
plt.hist(values, weights=pmf_values)
```

Графикке тақырып қосамыз.


```python
fig = plt.figure(figsize=(10,10))
plt.xticks(range(4))
plt.hist(values, weights=pmf_values)
fig.suptitle('3 тиын лақтыру үшін PMF', fontsize=15, y=0.92)
```

x және y осьтеріне атау береміз.


```python
fig = plt.figure(figsize=(10,10))
plt.xticks(range(4))
plt.hist(values, weights=pmf_values)
plt.xlabel('Елтаңба саны')
plt.ylabel('Ықтималдық')
fig.suptitle('3 тиын лақтыру үшін PMF', fontsize=15, y=0.92)
```

**Лабораториялық жұмыста берілетін барлық графиктерде тақырып және x, y осьтерінің атаулары міндетті болуы керек.**


### Биномдық үлестірімге қайта оралу

Төменде 3 сынақ үшін CDF мәндерін есептеу көрсетілген.


```python
values = np.array([])
cdf_values = np.array([])
for i in range(4):
    values = np.append(values, i)
    cdf_values = np.append(cdf_values, stats.binom.cdf(n=3, p=0.5, k=i))
```

CDF-ті бағандық немесе сызықтық графикпен көрсетуге болады. Қадамдық сызықтық график CDF табиғатына көбірек сәйкес келеді.


```python
fig = plt.figure(figsize=(10,10))
plt.xticks(range(4))
plt.plot(values, cdf_values,  drawstyle='steps-post', linestyle='-')
plt.xlabel('Елтаңба саны')
plt.ylabel('Ықтималдық')
fig.suptitle('3 тиын лақтыру үшін CDF', fontsize=15, y=0.92)
```


## Python тапсырмасы 1. NBA сериясы — тұрақты жеңіс ықтималдығы

Спортта биномдық ықтималдықтар плей-офф серияларының нәтижесін болжау үшін қолданылуы мүмкін.

New York Knicks NBA финалында Los Angeles Lakers командасымен **7 ойынға дейінгі серияда** ойнайды деп есептеңіз. Knicks-тің әр жеке ойында жеңу ықтималдығы:

$$p=0.40.$$

Серияны жеңудің төрт мүмкін ұзақтығы бар:
- 4 ойында жеңу;
- 5 ойында жеңу (4 жеңіс, 1 жеңіліс);
- 6 ойында жеңу (4 жеңіс, 2 жеңіліс);
- 7 ойында жеңу (4 жеңіс, 3 жеңіліс).

Әр сценарийде Knicks **соңғы ойынды міндетті түрде жеңуі керек**. Мысалы, серияны 5 ойында аяқтау үшін алғашқы 4 ойынның дәл 3-еуінде жеңіп, 5-ойында міндетті түрде жеңуі керек.

### Орындау керек
1. Knicks-тің серияны 4 ойында жеңу ықтималдығын есептеңіз.
2. 5 ойында жеңу ықтималдығын есептеңіз.
3. 6 ойында жеңу ықтималдығын есептеңіз.
4. 7 ойында жеңу ықтималдығын есептеңіз.
5. Осы төрт ықтималдықты қосып, Knicks-тің серияны жалпы жеңу ықтималдығын табыңыз.
6. Нәтижелерді Python арқылы тексеріңіз және қысқаша қорытынды жазыңыз.

**Нұсқау:** серия $n$ ойында аяқталса, алғашқы $n-1$ ойында Knicks дәл 3 рет жеңуі және соңғы ойында жеңуі керек.


```python
from math import comb

p = 0.40
q = 1 - p

series_probs = {}
for n in [4, 5, 6, 7]:
    prob = comb(n - 1, 3) * (p ** 4) * (q ** (n - 4))
    series_probs[n] = prob
    print(f"{n} ойында жеңу ықтималдығы: {prob:.6f}")

total_probability = sum(series_probs.values())
print(f"Жалпы серия жеңу ықтималдығы: {total_probability:.6f}")

# Қысқаша қорытынды: Knicks серияны жеңудің 4, 5, 6, 7 ойындағы ықтималдықтары
# бір-біріне тәуелді, өйткені соңғы ойында міндетті түрде жеңу керек.
```


## Python тапсырмасы 2. NBA сериясы — үйдегі және сырттағы ықтималдықтар әртүрлі

Ойындар келесі ретпен өтеді:
- 1–2 ойын: Los Angeles;
- 3–5 ойын: New York;
- 6–7 ойын: Los Angeles.

Knicks-тің жеңу ықтималдығы:

$$P(\text{жеңіс | Los Angeles})=0.30,$$

$$P(\text{жеңіс | New York})=0.60.$$

Knicks серияны 4, 5, 6 немесе 7 ойында жеңе алады. Әр жағдайда серияның соңғы ойыны міндетті түрде Knicks жеңісі болуы тиіс.

### Орындау керек
1. 4 ойында жеңудің барлық жарамды сценарийлерін жазыңыз және ықтималдығын есептеңіз.
2. 5 ойында жеңудің барлық сценарийлерін жазыңыз және ықтималдығын есептеңіз.
3. 6 ойында жеңудің барлық сценарийлерін жазыңыз және ықтималдығын есептеңіз.
4. 7 ойында жеңудің барлық сценарийлерін жазыңыз және ықтималдығын есептеңіз.
5. Барлық сценарийлерді қосып, серияны жалпы жеңу ықтималдығын табыңыз.
6. Python кодында барлық жарамды W/L тізбектерін автоматты түрде құрып, есептеуді тексеріңіз.

**Маңызды:** мұнда бір ғана тұрақты $p$ жоқ. Сондықтан қарапайым $Bin(n,p)$ моделін тікелей қолдануға болмайды; әр ойынның орналасуына сәйкес 0.30 немесе 0.60 ықтималдығын пайдаланыңыз.


```python
import itertools

locations = ['LA', 'LA', 'NY', 'NY', 'NY', 'LA', 'LA']
win_prob = {'LA': 0.30, 'NY': 0.60}

all_series = []
for n in [4, 5, 6, 7]:
    valid = []
    for seq in itertools.product(['W', 'L'], repeat=n - 1):
        if seq.count('W') != 3:
            continue
        prob = 1.0
        for i, outcome in enumerate(seq):
            if outcome == 'W':
                prob *= win_prob[locations[i]]
            else:
                prob *= 1 - win_prob[locations[i]]
        prob *= win_prob[locations[n - 1]]
        valid.append((seq + ('W',), prob))
    all_series.append((n, valid))
    print(f"\n{n} ойында жеңу мүмкін сценарийлері:")
    for seq, prob in valid:
        print(seq, f"=> {prob:.6f}")
    print(f"{n} ойында жеңу жалпы ықтималдығы: {sum(p for _, p in valid):.6f}")

total_probability = sum(prob for _, valid in all_series for _, prob in valid)
print(f"\nБарлық сценарийлерді қосқанда Knicks-тің серияны жалпы жеңу ықтималдығы: {total_probability:.6f}")
```


## Python тапсырмасы 3. Үш ойын сүйегінің PMF және CDF функциялары

Сабақта екі ойын сүйегінің қосындысы үшін PMF және CDF есептелді. Енді **үш әділ алтыжақты ойын сүйегін** лақтырыңыз.

$$X=D_1+D_2+D_3.$$

### Орындау керек
1. Үш ойын сүйегін лақтырғандағы барлық элементар нәтижелерді құрыңыз ($6^3=216$ нәтиже).
2. $X$ кездейсоқ шамасының барлық мүмкін мәндерін анықтаңыз.
3. Әр $x$ үшін PMF есептеңіз: $P(X=x)$.
4. PMF кестесін шығарыңыз.
5. Әр $x$ үшін CDF есептеңіз: $F(x)=P(X\le x)$.
6. CDF кестесін шығарыңыз.
7. PMF графигін салыңыз.
8. CDF-ті қадамдық график түрінде салыңыз.
9. Екі графикте де тақырып, x осі және y осі дұрыс белгіленуі керек.

**Нұсқау:** екі сүйек үшін екі кірістірілген `for` циклі қолданылса, үш сүйек үшін неше кірістірілген цикл қажет екенін ойлаңыз.


```python
three_dice = []
for i in range(1, 7):
    for j in range(1, 7):
        for k in range(1, 7):
            three_dice.append(i + j + k)

dice_values, counts = np.unique(np.array(three_dice), return_counts=True)
pmf = counts / np.sum(counts)
cdf = np.cumsum(pmf)

print("X мүмкін мәндері:", dice_values)
print("PMF:")
print(np.column_stack((dice_values, pmf)))
print("\nCDF:")
print(np.column_stack((dice_values, cdf)))

expected_value = np.sum(dice_values * pmf)
variance = np.sum((dice_values - expected_value) ** 2 * pmf)
std_dev = np.sqrt(variance)
print(f"\nE(X) = {expected_value:.4f}")
print(f"Var(X) = {variance:.4f}")
print(f"SD(X) = {std_dev:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(dice_values, pmf, width=0.8, color="steelblue")
axes[0].set_title("Үш ойын сүйегінің қосындысы үшін PMF")
axes[0].set_xlabel("X = D1 + D2 + D3")
axes[0].set_ylabel("Ықтималдық массасы")
axes[0].set_xticks(dice_values)

axes[1].step(dice_values, cdf, where="post", color="darkorange", linewidth=2)
axes[1].set_title("Үш ойын сүйегінің қосындысы үшін CDF")
axes[1].set_xlabel("X = D1 + D2 + D3")
axes[1].set_ylabel("Жинақталған ықтималдық")
axes[1].set_xticks(dice_values)

plt.tight_layout()
plt.show()
```
