### 2.2, №11(а)
**Шарт**: 
Пайлық инвестициялық қорлар компаниясының теңгерімді қорына тиесілі клиенттер үлесі 7% ($0.07$). Тек бір қордың үлестерін иеленетін клиент кездейсоқ таңдалса, оның теңгерімді қор үлестерін иелену ықтималдығы қандай?   
**Шешуі**:
Кестеде берілгендей, теңгерімді қордың үлесі $7\%$ құрайды. Деректер пайызбен берілгендіктен, кездейсоқ таңдалған клиенттің осы қорды иелену ықтималдығы тікелей осы үлеске тең:

```python
import matplotlib.pyplot as plt

p = 0.07
proportions = [p, 1 - p]
labels = [f'Теңгерімді\nP = {p:.2f}', f'Басқа оқиғалар\nP = {1 - p:.2f}']

fig, ax = plt.subplots(figsize=(7, 5))
wedges, texts, autotexts = ax.pie(
    proportions, labels=labels, colors=['#ff7f0e', '#1f77b4'],
    autopct='%1.0f%%', startangle=140, counterclock=False,
    explode=[0.08, 0], shadow=True,
    textprops={'fontsize': 11}, pctdistance=0.75,
    wedgeprops={'linewidth': 1.5, 'edgecolor': 'white'})
autotexts[0].set_color('white')
autotexts[0].set_fontweight('bo10ld')
autotexts[1].set_color('white')
autotexts[1].set_fontweight('bold')

ax.set_title('No11(а). Ықтималдықтың бөлінуі', fontsize=13, fontweight='bold', pad=14)
ax.text(0, -1.45, f'Қорытынды: P(Теңгерімді) = {p:.2f} = {p*100:.0f}%,  '
                  f'ал қалған {(1-p)*100:.0f}% — басқа оқиғалар',
        fontsize=10, ha='center', style='italic')
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```

### 2.3, №29(а)
**Есеп:** Тізбектелген екі әріптен ғана тұратын неше домен атауын жасауға болады?   Әріптермен бірге цифрларға (0–9) да рұқсат етілсе, ұзындығы екі таңба болатын неше атау бар?   
**Шешуі**:
1) Тек әріптер: Латын әліпбиінде 26 әріп бар. Әр орынды 26 тәсілмен таңдауға болатындықтан, көбейту ережесі бойынша:
   $$N_1 = 26 \times 26 = 676$$2) Әріптер мен цифрлар: Алфавитте 26 әріп + 10 цифр = 36 мүмкін таңба бар. Олай болса:$$N_2 = 36 \times 36 = 1296$$Жауабы:$$676$$ $$1296$$

```python
import matplotlib.pyplot as plt

categories = ['Тек әріптер\n(26 таңба)', 'Әріп + цифр\n(36 таңба)']
values = [26 ** 2, 36 ** 2]
colors = ['#4c72b0', '#55a868']

fig, ax = plt.subplots(figsize=(8, 5))
bars = ax.bar(categories, values, color=colors, width=0.45,
              edgecolor='black', linewidth=1.2)

for bar, v in zip(bars, values):
    ax.text(bar.get_x() + bar.get_width() / 2, v + 25,
            f'{v}', ha='center', fontsize=13, fontweight='bold')

ax.set_ylabel('Комбинациялар саны $N$', fontsize=12)
ax.set_xlabel('Таңба жиыны', fontsize=12)
ax.set_title('No29(а). 2 таңбалы домен атауларының саны', fontsize=13,
             fontweight='bold', pad=12)
ax.set_ylim(0, max(values) * 1.18)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.set_axisbelow(True)
ax.annotate(f'Айырмашылық:\n+{values[1]-values[0]} комбинация\n'
            f'(×{values[1]/values[0]:.2f})',
            xy=(1, values[1]), xytext=(0.42, values[1] * 0.75),
            fontsize=10, ha='center',
            arrowprops=dict(arrowstyle='->', color='#333333'))
plt.tight_layout()
plt.show()
```

### 2.4, №53
**Есеп:** $B \subset A$ болса, $P(B|A)$ шартты ықтималдығын табыңыз.

$B \subset A \implies A \cap B = B$
$P(B|A) = \frac{P(B)}{P(A)} = \frac{0.05}{0.6} = \frac{1}{12} \approx 0.0833$

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(4, 4))

# A (үлкен) және B (ішіндегі)
ax.add_patch(plt.Circle((0.5, 0.5), 0.4, color='#93c5fd', label='A'))
ax.add_patch(plt.Circle((0.5, 0.4), 0.18, color='#1d4ed8', label='B'))

ax.text(0.5, 0.7, 'A', fontsize=14, fontweight='bold', ha='center')
ax.text(0.5, 0.4, 'B', fontsize=12, fontweight='bold', color='white', ha='center')

plt.title(r'$P(B \mid A) = \frac{P(B)}{P(A)} = \frac{0.05}{0.6} = \frac{1}{12} \approx 0.0833$', fontsize=11)
plt.axis('off')
plt.show()
```
