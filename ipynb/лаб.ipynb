## №11. PMF және ықтималдық графигі

**Берілгені:** келесі автомобильдердің 45%-ы төрт цилиндрлі, 40%-ы алты цилиндрлі, 15%-ы сегіз цилиндрлі. $X$ — келесі автомобильдегі цилиндрлер саны.

**Шешуі:** PMF — әрбір мүмкін мәнге сәйкес ықтималдықтар кестесі:
$$P(X=4)=0.45,\quad P(X=6)=0.40,\quad P(X=8)=0.15.$$

Кемінде 6 цилиндр дегеніміз $X\ge6$: $0.40+0.15=0.55$. Ал 6-дан көп дегеніміз $X>6$: тек 8 цилиндр, сондықтан $0.15$. Төмендегі код PMF-ті сызықтық график және бағандық график ретінде көрсетеді.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb
import itertools

x_cyl = np.array([4, 6, 8])
pmf_cyl = np.array([0.45, 0.40, 0.15])
print('PMF:', dict(zip(x_cyl, pmf_cyl)))
print('P(X >= 6) =', pmf_cyl[1:].sum())
print('P(X > 6) =', pmf_cyl[2])

fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].plot(x_cyl, pmf_cyl, marker='o')
axes[0].set_title('Цилиндрлер санының PMF')
axes[0].set_xlabel('Цилиндрлер саны, X')
axes[0].set_ylabel('P(X = x)')
axes[0].set_xticks(x_cyl)
axes[1].bar(x_cyl, pmf_cyl, width=1.2)
axes[1].set_title('PMF ықтималдық гистограммасы')
axes[1].set_xlabel('Цилиндрлер саны, X')
axes[1].set_ylabel('Ықтималдық')
axes[1].set_xticks(x_cyl)
plt.tight_layout()
plt.show()
```

## №14. PMF шарттарын тексеру

**Берілгені:** $Y$ кездейсоқ шамасы 1, 2, 3, 4 немесе 5 мәндерін қабылдайды. Оның ықтималдығы $p(y)=ky$ формуласы арқылы берілген.

**Шешуі:** барлық ықтималдықтардың қосындысы 1-ге тең болуы керек:
$$k(1+2+3+4+5)=15k=1.$$
Осыдан $k=1/15$.

**Жауабы:**
- $P(Y\le3)=(1+2+3)/15=0.40$;
- $P(2\le Y\le4)=(2+3+4)/15=0.60$;
- $p(y)=y^2/50$ PMF бола алмайды, себебі оның қосындысы $55/50=1.1$, ал PMF үшін қосынды дәл 1 болуы қажет.

## №15. Екі платаның ақаулы болуы

**Берілгені:** бес платаның ішінен кездейсоқ екі плата таңдалады. 1 және 2-платалар ақаулы, қалған үшеуі жарамды. Барлық жұп саны $\binom{5}{2}=10$.

**Шешуі:** $X$ — таңдалған екі платаның ішіндегі ақаулы платалар саны. Екі плата да жарамды болатын 3 жұп, біреуі ақаулы болатын 6 жұп, екеуі де ақаулы болатын 1 жұп бар.

**Жауабы:**
$$P(X=0)=3/10=0.30,\quad P(X=1)=6/10=0.60,\quad P(X=2)=1/10=0.10.$$
CDF мәндері: $F(0)=0.30$, $F(1)=0.90$, $F(2)=1$. Толық түрде: $F(x)=0$ ($x<0$); $0.30$ ($0\le x<1$); $0.90$ ($1\le x<2$); $1$ ($x\ge2$).

## №18. Екі ойын сүйегінің максимум мәні

**Берілгені:** екі әділ алтыжақты сүйек лақтырылады. $M=\max(D_1,D_2)$ — екі түскен санның үлкені. Барлық элементар нәтиже саны $6\cdot6=36$.

**Шешуі:** $M\le m$ болуы үшін екі сүйектің де мәні $m$-нен үлкен болмауы керек. Сондықтан $P(M\le m)=m^2/36$. Осыдан PMF:
$$P(M=m)=P(M\le m)-P(M\le m-1)=\frac{m^2-(m-1)^2}{36}=\frac{2m-1}{36}.$$

**Жауабы:** $m=1,2,3,4,5,6$ үшін PMF формуласы $(2m-1)/36$. Төмендегі код CDF-ті қадамдық графикпен көрсетеді.

```python
m = np.arange(1, 7)
pmf_max = (2 * m - 1) / 36
cdf_max = np.cumsum(pmf_max)
print(np.column_stack((m, pmf_max, cdf_max)))
plt.step(m, cdf_max, where='post')
plt.title('Екі сүйек максимумының CDF')
plt.xlabel('M = max(D1, D2)')
plt.ylabel('F(m) = P(M <= m)')
plt.xticks(m)
plt.grid(alpha=0.3)
plt.show()
```

## №20. PMF-тен CDF-ке өту

**Берілгені:** семинарға 3 жұп және 2 жеке адам шақырылған. Әр жұп бір көлікпен бірге келеді, сондықтан жұптың екі адамы да бірге кешігеді немесе екеуі де уақытында келеді. Әр топтың кешігу ықтималдығы $0.4$.

**Шешуі:** әр жұп кешіксе, кешіккен адамдар санына 2 қосылады; жеке адам кешіксе, 1 қосылады. Сондықтан
$$X=2B_1+2B_2+2B_3+B_4+B_5,$$
мұндағы әр $B_i$ үшін $P(B_i=1)=0.4$. Барлық 32 топ нәтижесін есептеу арқылы PMF табылады, ал CDF — PMF мәндерінің жинақталған қосындысы.

**Жауабы:** кодта $P(2\le X\le6)=0.7776$ есептеледі және CDF графигі салынады.

```python
x20 = np.arange(9)
pmf20 = np.array([0.07776, 0.10368, 0.19008, 0.20736, 0.1728, 0.13824, 0.06912, 0.03072, 0.01024])
cdf20 = np.cumsum(pmf20)
print('PMF:', np.column_stack((x20, pmf20)))
print('P(2 <= X <= 6) =', cdf20[6] - cdf20[1])
plt.step(x20, cdf20, where='post')
plt.title('Кешіккен адамдар санының CDF')
plt.xlabel('Кешіккен адамдар саны, X')
plt.ylabel('F(x) = P(X <= x)')
plt.xticks(x20)
plt.grid(alpha=0.3)
plt.show()
```

## №23. Берілген CDF арқылы ықтималдықтарды табу

- $P(2<X<5)=F(4)-F(2)=0.92-0.39=0.53$.

## №29. Математикалық күтім, дисперсия және стандартты ауытқу

**Берілгені:** флеш-жад сыйымдылығы $X$ GB. Мәндер: $x=(1,2,4,8,16)$, ықтималдықтар: $p=(0.05,0.10,0.35,0.40,0.10)$.

**Шешуі:** математикалық күтім $E(X)=\sum xp(x)$, дисперсия $V(X)=\sum(x-E(X))^2p(x)$, стандартты ауытқу $\sigma=\sqrt{V(X)}$.

**Жауабы:** $E(X)=6.45$; $E(X^2)=57.25$; $V(X)=15.6475$. Қысқаша формула да дәл осыны береді: $57.25-6.45^2=15.6475$. Сондықтан $\sigma=\sqrt{15.6475}=3.9557$ GB.

## №31. Дисперсия және стандартты ауытқу

**Берілгені:** жолаушылар саны $Y=45,46,\ldots,55$ және әр мәннің ықтималдығы кестеде берілген.

**Шешуі:** кестедегі барлық ықтималдықтарды пайдаланып, $E(Y)=\sum yp(y)$ және $V(Y)=\sum(y-E(Y))^2p(y)$ есептеледі. Нәтижесінде $E(Y)=48.84$, $V(Y)=4.4944$, сондықтан стандартты ауытқу:
$$\sigma_Y=\sqrt{V(Y)}=\sqrt{4.4944}=2.12.$$

Бір стандартты ауытқу шегі:
$$48.84-2.12\le Y\le48.84+2.12,$$
яғни бүтін мәндер үшін $Y=47,48,49,50$. Олардың ықтималдықтарын қосамыз: $0.12+0.14+0.25+0.17=0.68$.

**Жауабы:** $V(Y)=4.4944$, $\sigma_Y=2.12$, $P(|Y-E(Y)|\le\sigma_Y)=0.68$.

## Python 1. NBA сериясы — бірдей жеңіс ықтималдығы

**Берілгені:** Knicks командасы 7 ойынға дейінгі серияда ойнайды. Әр ойында жеңу ықтималдығы $p=0.40$, жеңілу ықтималдығы $q=1-p=0.60$. Серия $n$ ойында аяқталса, соңғы ойында міндетті түрде жеңіс болуы керек, ал алғашқы $n-1$ ойында дәл 3 жеңіс болуы тиіс.

**Шешуі:** соңғы жеңісті бөлек алып, алғашқы ойындардан 3 жеңісті таңдаймыз:
$$P_n=\binom{n-1}{3}p^4q^{n-4}.$$
Келесі код әр серия ұзақтығының ықтималдығын есептеп, нәтижелерді қосады.

```python
p = 0.40
series_probs = {n: comb(n - 1, 3) * p**4 * (1-p)**(n-4) for n in range(4, 8)}
for n, probability in series_probs.items():
    print(f'{n} ойында: {probability:.6f}')
print(f'Жалпы ықтималдық: {sum(series_probs.values()):.6f}')
```

## Python 2. Үйдегі және сырттағы ойындар

**Берілгені:** ойындардың реті: 1–2 ойын — Los Angeles, 3–5 ойын — New York, 6–7 ойын — Los Angeles. Knicks-тің LA алаңындағы жеңіс ықтималдығы $0.30$, NY алаңындағы жеңіс ықтималдығы $0.60$.

**Шешуі:** бұл жерде бір ғана тұрақты $p$ қолдануға болмайды. Код барлық жарамды `W/L` тізбектерін құрады: соңғы ойын әрқашан `W`, ал оған дейінгі ойындарда дәл үш `W` болуы керек. Әр тізбектің ықтималдығы ойын өтетін орынға сәйкес көбейтіледі.

```python
locations = ['LA', 'LA', 'NY', 'NY', 'NY', 'LA', 'LA']
win_prob = {'LA': 0.30, 'NY': 0.60}
totals = {}

for n in range(4, 8):
    valid_scenarios = []
    for prefix in itertools.product(('W', 'L'), repeat=n - 1):
        # Серия n ойында аяқталуы үшін соңғы ойын W,
        # ал алғашқы n-1 ойында дәл 3 жеңіс болуы керек.
        if prefix.count('W') != 3:
            continue

        sequence = prefix + ('W',)
        probability = 1.0
        for game_number, result in enumerate(sequence):
            location = locations[game_number]
            probability *= (
                win_prob[location]
                if result == 'W'
                else 1 - win_prob[location]
            )
        valid_scenarios.append((sequence, probability))

    totals[n] = sum(probability for _, probability in valid_scenarios)
    print(f'\n{n} ойында жеңу сценарийлері:')
    for sequence, probability in valid_scenarios:
        print(''.join(sequence), f'-> {probability:.6f}')
    print(f'{n} ойындағы жалпы ықтималдық: {totals[n]:.6f}')

print(f'\nЖалпы серияны жеңу ықтималдығы: {sum(totals.values()):.6f}')
```

## Python 3. Үш сүйек үшін PMF және CDF

**Берілгені:** үш әділ алтыжақты сүйек лақтырылады. Олардың қосындысын $X=D_1+D_2+D_3$ деп белгілейміз. Әр сүйектің мәндері 1-ден 6-ға дейін, сондықтан элементар нәтижелердің жалпы саны $6^3=216$.

**Шешуі:** үш кірістірілген цикл барлық 216 нәтижені құрады. Әр қосындының жиілігі табылады; PMF жиілікті 216-ға бөлу арқылы есептеледі:
$$P(X=x)=\frac{\text{X=x болатын нәтижелер саны}}{216}.$$
CDF PMF мәндерінің жинақталған қосындысы арқылы табылады:
$$F(x)=P(X\le x).$$
Код PMF және CDF кестелерін шығарады, сондай-ақ екі график салады.

```python
three_dice = [i + j + k for i in range(1, 7) for j in range(1, 7) for k in range(1, 7)]
values, counts = np.unique(three_dice, return_counts=True)
pmf = counts / len(three_dice)
cdf = np.cumsum(pmf)
print('Элементар нәтижелер саны:', len(three_dice))
print('PMF кестесі:')
print(np.column_stack((values, pmf)))
print('CDF кестесі:')
print(np.column_stack((values, cdf)))
print(f'E(X) = {np.sum(values * pmf):.4f}')
print(f'Var(X) = {np.sum((values - np.sum(values * pmf))**2 * pmf):.4f}')
print(f'SD(X) = {np.sqrt(np.sum((values - np.sum(values * pmf))**2 * pmf)):.4f}')

fig, axes = plt.subplots(1, 2, figsize=(13, 4))
axes[0].bar(values, pmf, color='steelblue')
axes[0].set_title('Үш сүйек қосындысының PMF')
axes[0].set_xlabel('X = D1 + D2 + D3')
axes[0].set_ylabel('P(X = x)')
axes[0].set_xticks(values)
axes[1].step(values, cdf, where='post', color='darkorange', linewidth=2)
axes[1].set_title('Үш сүйек қосындысының CDF')
axes[1].set_xlabel('X = D1 + D2 + D3')
axes[1].set_ylabel('F(x) = P(X <= x)')
axes[1].set_xticks(values)
axes[1].grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## Қорытынды

PMF — дискретті кездейсоқ шаманың әрбір нақты мәнінің ықтималдығы. CDF — $X$ шамасының берілген мәннен аспау ықтималдығы, яғни $F(x)=P(X\le x)$. Математикалық күтім шаманың орташа мәнін көрсетеді, дисперсия мәндердің орташа мәннен қаншалықты шашырағанын өлшейді, ал стандартты ауытқу дисперсияның квадрат түбірі болып табылады.
